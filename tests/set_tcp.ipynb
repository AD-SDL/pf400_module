```python
from pf400_interface.pf400 import PF400
```

```python
from madsci.common.types.location_types import LocationArgument
```

```python
pf400 = PF400(host="rplpf400.cels.anl.gov")
pf400.initialize_robot()
pf400.get_robot_movement_state()
```

```python
pf400.open_gripper(130)
```

```python
rotation_deck = LocationArgument(
    location_name="rotation_deck",
    location=[63.063, -28.717, 113.020, -82.561, 71.348, 988.948],
)
camera_deck = LocationArgument(
    location_name="camera_deck",
    location=[105.024, 22.029, 69.970, -0.551, 73.780, 991.020],
)
```

```python
camera = pf400.rotate_yaw(camera_deck.representation, -90)
camera_wide = LocationArgument(
    location_name="camera_wide",
    location=[104.961, 59.382, 36.824, -94.872, 122.180, 990.609],
)
print(camera_wide)
```

```python
pf400.send_command("tool")
```

```python
pf400.move_joint(camera_deck.representation)
```

```python
pf400.transfer(
    source=camera_wide,
    target=camera_deck,
    rotation_deck=rotation_deck,
    source_plate_rotation="wide",
    target_plate_rotation="narrow",
)
```

```python
cart = pf400.joint_to_cart(rotation_deck.representation)
print(cart)
pf400.move_with_rotation(rotation_deck.representation, -90.0)
```

```python
rot = pf400.rotate_yaw(rotation_deck.representation, 90)
rotation_deck_wide = LocationArgument(location_name="rotation_deck_wide", location=rot)
```

```python
pf400.transfer(
    rotation_deck_wide,
    rotation_deck_wide,
    source_plate_rotation="wide",
    target_plate_rotation="wide",
)
```

```python
pf400.send_command("tool 0 1 146.5 0 0 0")
pf400.send_command("StoreFile")
```

```python
pf400.disconnect()
```
